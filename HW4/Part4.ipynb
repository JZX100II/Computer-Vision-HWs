```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

gray = cv2.imread("MRI.jfif", cv2.IMREAD_GRAYSCALE)

if gray is None:
    raise FileNotFoundError("image not found")

# Apply OpenCV colormaps for pseudocolor visualization
colored_jet = cv2.applyColorMap(gray, cv2.COLORMAP_JET)
colored_hot = cv2.applyColorMap(gray, cv2.COLORMAP_HOT) # thermal style
# colored_cool = cv2.applyColorMap(gray, cv2.COLORMAP_COOL)

# Convert to RGB for Matplotlib display
gray_rgb = cv2.cvtColor(gray, cv2.COLOR_GRAY2RGB)
colored_jet_rgb = cv2.cvtColor(colored_jet, cv2.COLOR_BGR2RGB)
colored_hot_rgb = cv2.cvtColor(colored_hot, cv2.COLOR_BGR2RGB)

# Display
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(gray_rgb)
plt.title("Original Grayscale (MRI)")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(colored_jet_rgb)
plt.title("Pseudocolor: COLORMAP_JET")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(colored_hot_rgb)
plt.title("Pseudocolor: COLORMAP_HOT")
plt.axis("off")

plt.tight_layout()
plt.show()
```
